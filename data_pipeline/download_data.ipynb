3. Place it at `~/.kaggle/kaggle.json` (Linux/Mac) or `C:\Users\<you>\.kaggle\kaggle.json` (Windows).

```python
import sys
import zipfile
import logging

logging.basicConfig(level=logging.INFO, format="%(asctime)s | %(levelname)s | %(message)s")
logger = logging.getLogger("download")


def download_dataset() -> None:
    """Download and extract the resume dataset from Kaggle."""
    if RAW_RESUME_CSV.exists():
        logger.info(f"Dataset already present at {RAW_RESUME_CSV}, skipping download.")
        return

    try:
        from kaggle.api.kaggle_api_extended import KaggleApi
    except ImportError:
        logger.error("The 'kaggle' package isn't installed. Run: pip install kaggle")
        sys.exit(1)
    except OSError as e:
        logger.error(
            "Kaggle API credentials not found or invalid. See the setup "
            "instructions above.\nOriginal error: %s", e
        )
        sys.exit(1)

    api = KaggleApi()
    api.authenticate()

    logger.info(f"Downloading '{KAGGLE_DATASET_SLUG}' to {RAW_DATA_DIR} ...")
    api.dataset_download_files(KAGGLE_DATASET_SLUG, path=str(RAW_DATA_DIR), unzip=False)

    zip_path = RAW_DATA_DIR / "resume-dataset.zip"
    if zip_path.exists():
        logger.info(f"Extracting {zip_path} ...")
        with zipfile.ZipFile(zip_path, "r") as zf:
            zf.extractall(RAW_DATA_DIR)
        zip_path.unlink()

    if RAW_RESUME_CSV.exists():
        logger.info(f"Success. Raw data available at {RAW_RESUME_CSV}")
    else:
        logger.warning(
            "Download finished but Resume.csv wasn't found where expected. "
            "Check data/raw/ and update RAW_RESUME_CSV if the filename differs."
        )


download_dataset()
```
